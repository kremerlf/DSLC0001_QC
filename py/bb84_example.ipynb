```python
import pandas as pd

from bb84 import Qb, Comms, User
```

```python
runs = 1
bits = 64

tempo = [] # save results and get statistics
for i in range(runs):
    Alice = User()
    Bob = User()
    
    # random measures and send
    Alice.stream_gen(bits)
    stream = Comms.send(Alice)
    
    # measures the sended stream
    Bob.measure(stream)
    
    # open Comms basis check and individual key generation
    Bob.key_gen(Alice)
    Alice.key_gen(Bob)

    
    tempo.append([1 if Alice.key == Bob.key else 0, 
                len(Bob.key)])


ab = pd.DataFrame(tempo, columns = ['equal_key', 'key_len'])
del tempo
```
